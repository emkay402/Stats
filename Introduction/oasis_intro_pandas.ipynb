## 4. Structure of a DataFrame and how to access elements

A DataFrame is a collection of **columns**, where each column is a **Series** (a labelled list of values). Every row has an **index label** (here `0, 1, 2, ...`) and every column has a **name**.

````python
print(type(df))
print(type(df["Age"]))
````

````python
df["Age"].head()
````

The numbers on the left are the index labels. There are three common ways to pick out a single value. As an example, we will get the **age of the first participant** with each of them.

### 4a. Column name first, then index

`df["Age"]` gives the whole column (a Series), and `[0]` then picks the entry with **index label** 0.

````python
df["Age"][0]
````

````python
df["Age"][0:5]   # a slice of the column
````

### 4b. By position, like a matrix: `.iloc[row, column]`

`iloc` stands for **integer location**. You give *row number, column number*, counting from **0**, exactly as in a matrix. `Age` is the 4th column, so its position is 3.

````python
df.iloc[0, 3]
````

````python
df.iloc[0:5, 2:5]   # rows 0-4 and columns 2-4: the end of the range is EXCLUDED
````

````python
df.iloc[0]          # a whole row (the first participant)
````

### 4c. Row first, then column name: `.loc[row, column name]`

`loc` stands for **label location**. You give the row's *index label* and the column's *name*.

````python
df.loc[0, "Age"]
````

````python
df.loc[0:4, ["Age", "MMSE"]]   # careful: with .loc the end of the range is INCLUDED
````

````python
# All three ways return the same value:
print(df["Age"][0], df.iloc[0, 3], df.loc[0, "Age"])
````

### Summary

| Method | You provide | Counting | End of a slice |
|---|---|---|---|
| `df["Age"][0]` | column name, then index label | by label | label-based |
| `df.iloc[0, 3]` | row position, column position | from 0 | **excluded** |
| `df.loc[0, "Age"]` | row label, column name | by label | **included** |

### A pitfall: labels and positions are not always the same

Right now, the label of each row equals its position, so the three methods agree. As soon as we **filter** the table, this changes.

````python
oldest = df[df["Age"] >= 90]   # keep only participants aged 90 or older
oldest.head(3)
````

The index labels of the filtered table are no longer `0, 1, 2`. Now what happens?

````python
print(oldest.iloc[0, 3])        # first row *by position*: works
print(oldest.loc[oldest.index[0], "Age"])   # first row *by label*: works, we looked up the label
````

````python
try:
    print(oldest.loc[0, "Age"])  # label 0 was filtered out
except KeyError as e:
    print("KeyError: there is no row with label", e)
````

> **Rule of thumb:** use `.iloc` when you mean "the n-th row" and `.loc` when you mean "the row with this label". Both are safer than the chained form `df["Age"][0]`, which is fine for quick looks.

## 6. Mapping: replacing text with numbers

Many analyses need numeric inputs. Mapping means replacing each value with a chosen replacement. In pandas we do it with a **dictionary** and `.map()`:

```python
{old_value: new_value, ...}
```

### 6a. Nominal variable: `M/F`

Sex is **nominal** (no order), stored as text. We map `F -> 0` and `M -> 1`.

````python
sex_map = {"F": 0, "M": 1}

df["Sex_code"] = df["M/F"].map(sex_map)
df[["M/F", "Sex_code"]].head()
````

Always **verify** a mapping. A cross-tabulation should show that every `F` became 0 and every `M` became 1:

````python
pd.crosstab(df["M/F"], df["Sex_code"])
````

**Watch out:** `.map()` turns every value that is *not* in the dictionary into `NaN`. A typo silently destroys data:

````python
bad_map = {"F": 0, "m": 1}          # lower-case "m" does not match "M"
print(df["M/F"].map(bad_map).isna().sum(), "values were lost")
````

Good habit: after mapping, check that you introduced no new missing values.

````python
assert df["Sex_code"].isna().sum() == df["M/F"].isna().sum()
print("No values were lost.")
````

### 6b. Ordinal variables: `CDR`, `Educ`, `SES`

For ordinal variables the numbers should **respect the order**. Three situations come up:

**(i) Turning odd codes into ranks.** `CDR` has the values 0, 0.5, 1, 2, so the steps are not equal. Mapping to ranks 0, 1, 2, 3 gives a clean ordinal scale. (Whether to treat the gaps as equal is a decision you should make consciously.)

````python
cdr_rank_map = {0.0: 0, 0.5: 1, 1.0: 2, 2.0: 3}

df["CDR_rank"] = df["CDR"].map(cdr_rank_map)
pd.crosstab(df["CDR"], df["CDR_rank"])
````

Note that `NaN` stays `NaN`: missing values are not in the dictionary, so they are not assigned a number. (`crosstab` hides them by default.)

**(ii) Attaching readable labels.** Mapping can also go from numbers to text, which makes tables and plots easier to read:

````python
cdr_label_map = {0.0: "nondemented", 0.5: "very mild", 1.0: "mild", 2.0: "moderate"}
educ_label_map = {1: "< high school", 2: "high school grad", 3: "some college",
                  4: "college grad", 5: "beyond college"}

df["CDR_label"] = df["CDR"].map(cdr_label_map)
df["Educ_label"] = df["Educ"].map(educ_label_map)
df[["CDR", "CDR_label", "Educ", "Educ_label"]].dropna().head()
````

**(iii) Reversing a scale.** For `SES`, **1 = highest** and **5 = lowest** status, the opposite of `Educ`, where 5 = most education. That is confusing: a positive relationship between status and anything would show up as a *negative* correlation. We can create a reversed version where higher means higher status:

````python
ses_reverse_map = {1: 5, 2: 4, 3: 3, 4: 2, 5: 1}

df["SES_reversed"] = df["SES"].map(ses_reverse_map)
pd.crosstab(df["SES"], df["SES_reversed"])
````

> **Question:** Compare the correlation of `Educ` with `SES` and with `SES_reversed`. What changes, and what stays the same?

````python
print("Educ vs SES:         ", round(df["Educ"].corr(df["SES"]), 3))
print("Educ vs SES_reversed:", round(df["Educ"].corr(df["SES_reversed"]), 3))
````

Only the **sign** flips; the strength is identical. This shows why you must know how a variable is coded before you interpret a result.

### 6c. Keeping a dictionary of your mappings

Months later you will not remember whether `M` was 0 or 1. Collect all mappings in **one dictionary** that serves as a codebook:

````python
mappings = {
    "Sex_code":     {"F": 0, "M": 1},
    "CDR_rank":     {0.0: 0, 0.5: 1, 1.0: 2, 2.0: 3},
    "CDR_label":    {0.0: "nondemented", 0.5: "very mild", 1.0: "mild", 2.0: "moderate"},
    "Educ_label":   {1: "< high school", 2: "high school grad", 3: "some college",
                     4: "college grad", 5: "beyond college"},
    "SES_reversed": {1: 5, 2: 4, 3: 3, 4: 2, 5: 1},
}

pprint(mappings)
````

````python
# Look up one mapping whenever you forget it:
mappings["Sex_code"]
````

A mapping can be **inverted** to go back from the numbers to the original values:

````python
inverse_sex_map = {number: text for text, number in mappings["Sex_code"].items()}
print(inverse_sex_map)

df["Sex_decoded"] = df["Sex_code"].map(inverse_sex_map)
print("Round trip successful:", (df["Sex_decoded"] == df["M/F"]).all())
````

## Extra: one-hot encoding

### The problem with numeric codes for nominal variables

Suppose a variable has three **unordered** categories, e.g. blood type A, B, O. If we code them 0, 1, 2, then any model will treat "O" as twice "B" and "B" as halfway between "A" and "O". That is meaningless for a nominal variable.

### The solution

**One-hot encoding** creates one new 0/1 column per category. A row has a `1` in the column of its category and `0` everywhere else.

In pandas: `pd.get_dummies()`. Our dataset has no unordered variable with more than two categories, so we use `CDR_label` purely as a demonstration.

````python
onehot = pd.get_dummies(df["CDR_label"], prefix="CDR", dtype=int)
onehot.head(10)
````

````python
pd.concat([df[["CDR_label"]], onehot], axis=1).dropna().head()
````

Every row (with a rating) should contain exactly one `1`:

````python
row_sums = onehot.sum(axis=1)
print(row_sums.value_counts())
````

> **Question:** Some rows sum to **0**. Which ones, and why? What would go wrong if we didn't notice this?

Rows with a missing `CDR` get **all zeros**. By default, `get_dummies` does not create a "missing" category, so a participant with *no rating* looks the same as one that is "in none of the categories". You can handle this in two ways:

````python
# Option 1: add an explicit "missing" column
pd.get_dummies(df["CDR_label"], prefix="CDR", dtype=int, dummy_na=True).head(5)
````

````python
# Option 2: drop the rows with missing values first
pd.get_dummies(df["CDR_label"].dropna(), prefix="CDR", dtype=int).head(5)
````

### The dummy variable trap: `drop_first`

If you know a participant is **not** in 3 of 4 categories, you automatically know they are in the fourth. The 4th column is redundant, and in a regression model with an intercept it causes perfect **multicollinearity**. The usual fix is to drop one column, which becomes the **reference category**:

````python
pd.get_dummies(df["CDR_label"].dropna(), prefix="CDR", dtype=int, drop_first=True).head()
````

Note that `drop_first=True` drops the first category *alphabetically* (here "mild"). If you want a specific reference group (e.g. "nondemented"), drop that column yourself.

### When to use what?

| Variable type | Example | Encoding |
|---|---|---|
| Nominal, 2 categories | `M/F` | a single 0/1 column (what we did in 6a) |
| Nominal, 3+ categories | e.g. brain region, diagnosis group | one-hot encoding |
| Ordinal | `CDR`, `Educ`, `SES` | rank mapping (as in 6b), *or* one-hot if you do not want to assume equal spacing |
| Continuous | `Age`, `nWBV` | no encoding needed |
